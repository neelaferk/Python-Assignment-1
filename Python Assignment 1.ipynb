## String Concatenation :

```python
name = input("Enter Your Name: ")

string1 = "Hello "
result = string1 + name

print(result)

string3 = " Welcome to Python programing."

result = result + string3

print(result)
```

## String Slicing And Indexing :

```python
# Print the first character of the string

s = ("Hello zara Welcome to Python programing.")
s[0]


```

```python
# Print the last character of string

s = ("Hello zara Welcome to Python programing.")
s[-2]

```

```python
# Print the First 5 characters of string

s = ("Hello zara Welcome to Python programing.")
s[0:5]

```

```python
# print the last 11 characters of string

s = ("Hello zara Welcome to Python programing.")
s[-11:-1]
```

```python
# print the string in reverse

s = ("Hello zara Welcome to Python programing.")
s[-2 : -12 : -1]
```

```python
# using slicing and print the word Python from the existing string

s = ("Hello zara Welcome to Python programing.")
s[22 : 28]

```

## String Methods:

```python
# Convert the sentences to uppercase

s = "python beginner tutorial,"
s.upper()

```

```python
# Convert the sentences to Lowercase

s = "python beginner tutorial,"
s.lower()



```

```python
# use capitalize and return the sentence to the original input form

s = "python beginner tutorial,"
s.capitalize()

```

```python
# Count the total number of occurenences of character T in the string

s = "python beginner tutorial,"
s.count("t")
```

```python
# Replace python with Data analytics

str = "Python beginner tutorial"
str.replace("Python","Data Analytics")
```

## Tuples (Creation, modification and Access)

```python
# Concatenate the two tuples and store it in "t_combine"

t1 = (10,20,30)
t2 = (40,50,60)
t_combine = t1 + t2
print(t_combine)
```

```python
# Repeat the elements of t_combine 3 times

t1 = (10,20,30)
t2 = (40,50,60)
t_combine = t1 + t2
t_repeat = t_combine * 3
print(t_repeat)


```

```python
# Access the 3rd element from t_combine

t_combine[2]
```

```python
# Acess the first 3 elements fromm t_combine

t_combine
t_combine[0 : 3]
```

```python
# Acess the last 3 elements fromm t_combine

t_combine
t_combine[3 : ]


```
